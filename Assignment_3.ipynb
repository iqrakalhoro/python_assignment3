````python
my_list = [1, 2, 3, 4, 5]  # Create a list of numbers
print(my_list)
````

````python
print(my_list[0])  # Access the first element (index 0)
print(my_list[-1]) # Access the last element (index -1)
````

````python
print(my_list[1:4])  # Get a sublist from index 1 to 3 (exclusive of 4)
````

````python
my_list[2] = 10  # Change the value at index 2 to 10
print(my_list)
````

````python
my_list.append(6)  # Add an element to the end
my_list.insert(2, 7)  # Insert an element at index 2
my_list.remove(4)  # Remove the first occurrence of the value 4
print(my_list)
````

````python
squares = [x**2 for x in range(1, 6)]  # Create a list of squares using list comprehension
print(squares)
````

````python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]  # A list of lists (a matrix)
print(matrix[1][2])  # Access the element at row 1, column 2
````

````python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
combined_list = list1 + list2  # Concatenate two lists
print(combined_list)
````

````python
empty_tuple = ()  # An empty tuple
numbers_tuple = (1.0, 9.9, 10)  # Tuple with numeric objects
names_tuple = ('Casey', 'Darin', 'Bella', 'Mehdi')  # Tuple with string objects
mixed_tuple = ('10', 101, True)  # Tuple with mixed data types
````

````python
print(names_tuple[1:3])  # Get elements from index 1 to 2 (exclusive of 3) - Output: ('Darin', 'Bella')
````

````python
# Using curly braces:
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
print(my_dict)

# Using dict():
another_dict = dict(name="Bob", age=25, city="London")
print(another_dict)
````

````python
print(my_dict["name"])  # Accessing the value associated with the key "name"
print(my_dict.get("age")) # Using get() method (returns None if key not found)
````

````python
my_dict["occupation"] = "Engineer"  # Adding a new key-value pair
my_dict["age"] = 31  # Updating the value of an existing key
print(my_dict)
````

````python
del my_dict["city"]  # Removing the key-value pair with key "city"
print(my_dict)

removed_value = my_dict.pop("age")  # Removing and returning the value
print(removed_value)
print(my_dict)
````

````python
if "name" in my_dict:
    print("Name is a key in the dictionary")
````

````python
# Iterating through keys:
for key in my_dict:
    print(key)

# Iterating through values:
for value in my_dict.values():
    print(value)

# Iterating through key-value pairs:
for key, value in my_dict.items():
    print(key, value)
````

````python
squares_dict = {x: x**2 for x in range(1, 6)}
print(squares_dict)
````

````python
my_set = {1, 2, 3, 4, 5}  # Using curly braces
another_set = set([3, 4, 5, 6, 7])  # Using the set() constructor
print(my_set)
print(another_set)
````

````python
my_set.add(6)  # Adding an element
my_set.remove(3)  # Removing an element (raises KeyError if not found)
my_set.discard(3)  # Removing an element (does nothing if not found)
print(my_set)
````

````python
set1 = {1, 2, 3}
set2 = {3, 4, 5}

union_set = set1 | set2  # Union of sets
intersection_set = set1 & set2  # Intersection of sets
difference_set = set1 - set2  # Difference of sets
print(union_set)
print(intersection_set)
print(difference_set)
````

````python
frozen_set = frozenset([1, 2, 3, 4, 5])
print(frozen_set)
````
